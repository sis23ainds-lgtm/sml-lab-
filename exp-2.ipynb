```python
cross_tab=pd.crosstab(data["Satisfaction_Level"],data["Repeat_Purchase"])
```

```python
cross_tab
```

```python
cross_tab.plot(kind='bar', figsize=(8,5))
plt.title('Repeat Purchase by Satisfaction Level')
plt.xlabel('Satisfaction Level')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.show()
```

```python
cross_tab.plot(kind='bar', stacked=True, figsize=(8,5), color=['blue','green'])
plt.title('Repeat Purchase by Satisfaction Level (Stacked)')
plt.xlabel('Satisfaction Level')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.show()
```
